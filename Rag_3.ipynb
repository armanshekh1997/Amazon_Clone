```python
import chromadb
```

```python
from sentence_transformers import SentenceTransformer
```

```python
## creates embedding model
model=SentenceTransformer('all-MiniLM-L6-v2')
```

```python
##create chromadb client- this client is our connection/interface to chromadb
###python application>>client>>chromadb
client=chromadb.Client()
## amazon.com/user_return
##amazon.com/payment
## amazon.com/accounting
## though the client we can create collection>>>get the collection >>delete the collection>> add documet>>search collection
```

```python
##create collection
collection=client.create_collection(name='my_collection')
##container which store or documents,embedding, IDs and metadata
##client.get_colection()
## client.delete_collection()
```

```python
##database and table
##datrabase>>table>>rows
##collection as a table in a traditional database
## cgromadb >>collection>>documen+embedding+metadata
##HR document/Finance documents/IT documents
##collection cotain
documents=[
    "package has not arrivedafter 2 weeks",
    "i was charged twicefor the same order",
    "wrong item was delevered to me",
    "i can not reset my account password"
]
```

```python
## genetare embedding
embeddings = model.encode(documents)
print("number of documents:",len(documents))
print("number of embeddings:",len(embeddings))
```
