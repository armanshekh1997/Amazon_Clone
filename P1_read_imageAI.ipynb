# Objectiive : Upload random image and asked the gemini (AI) to look the image and answer the question

1) What is this image all about?
2) Describe everything you see?
3) Is there any animal?
4) How many people are there?

# FLOW:
user--->clicks "choose file" ---> sellects an image---> python reads the image---> image send to gemini-->gemini analyse it--->returns the description --->python print the result

```python
from google.colab import userdata
api_key = userdata.get("GEMINI_API_KEY")

from google import genai

client = genai.Client(api_key=api_key)
```

```python
from google.colab import files  # because we want user to upload a file
from google.genai import types # to package the image before sending it to gemini(model)
```

```python
uploaded = files.upload()
filename = list(uploaded.keys())[0]
image_bytes = uploaded[filename]

response = client.models.generate_content(
    model="gemini-3.8-flash",

    contents = ["describe this image in detail. List everythings you can see",
                types.Part.from_bytes(data=image_bytes,mime_type="image/jpeg"),
                ],

)

print(response.text)
print("\n Input Token :", response.usage_metadata.prompt_token_count)
print("\n Output Token :", response.usage_metadata.candidates_token_count)
```
